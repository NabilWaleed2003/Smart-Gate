```python
import torch

# ==========================================
# 1. Checkpoint path
# ==========================================

checkpoint_path = (
    "/content/drive/MyDrive/Smart Gate/Models/Yolo11m_truck_best.pt"
)


# ==========================================
# 2. Load checkpoint
# ==========================================

checkpoint = torch.load(
    checkpoint_path,
    map_location="cuda" if torch.cuda.is_available() else "cpu"
)

print("Best epoch:", checkpoint["epoch"])
print("Best val loss:", checkpoint["val_loss"])


# ==========================================
# 3. Recreate model
# ==========================================

trained_model = Build_model.TruckDetectionModel(
    model_weights_path="yolo11m.pt"
)


# ==========================================
# 4. Load trained weights
# ==========================================

trained_model.load_state_dict(
    checkpoint["model_state_dict"]
)


# ==========================================
# 5. Move to device
# ==========================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

trained_model = trained_model.to(device)
trained_model.eval()

verify_dm = data_module.TruckDataModule(
    data_dir=data_dir,
    batch_size=32,
    num_workers=0
)

verify_dm.setup()


# ==========================================
# 6. Evaluation
# ==========================================



# Pass the model object itself
results_pt = evaluate.per_class_acc_and_conf_matrix(trained_model, verify_dm)
size_pt = evaluate.get_model_size_mb(checkpoint_path)
time_pt = evaluate.measure_inference_time(trained_model)
```

## Consuming Model

```python
%%writefile "/content/drive/MyDrive/Smart Gate/Deployment/client_camera.py"
import os
import requests
import time

base_url = 'http://localhost:8000'
endpoint = '/api/v1/process_frame'
full_url = base_url + endpoint

def check_gate_access(url, image_path, verbose=True):
    if not os.path.exists(image_path):
        print(f"❌ Error: Image {image_path} not found.")
        return None

    with open(image_path, "rb") as image_file:
        # Include filename and MIME type so FastAPI accepts the upload
        files = {'file': (os.path.basename(image_path), image_file, 'image/jpeg')}

        try:
            response = requests.post(url, files=files)
            status_code = response.status_code

            if status_code == 200:
                server_decision = response.json()
                if verbose:
                    print(f"\n--- Checking Image: {os.path.basename(image_path)} ---")

                action = server_decision.get("action")
                if action == "open_gate":
                    truck_count = server_decision.get("trucks_in_frame")
                    print(f"🟢 [HARDWARE SIGNAL]: OPENING GATE! ({truck_count} New Truck(s) Detected)")
                else:
                    print("🔴 [HARDWARE SIGNAL]: KEEP GATE CLOSED. (No new trucks or nothing detected)")

                return server_decision
            else:
                print(f"❌ Server returned an error: {status_code}")
                # Print error details from the server if present
                print(f"Details: {response.text}")
                return None
        except requests.exceptions.ConnectionError:
            print("❌ Error: Could not connect to the API. Is the server running?")
            return None

if __name__ == "__main__":
    # Path to your test image
    img = "/content/drive/MyDrive/Smart Gate/Data/processed/Trucks/test/images/100_resized_jpg.rf.9a6e56b12127d9b7ae25529613a3b5f6.jpg"
    image_files = [img, img]

    print(f"Starting Smart Gate Camera Simulation targeting: {full_url}")
    print("=" * 60)

    for img_path in image_files:
        result = check_gate_access(full_url, img_path)
        time.sleep(1)
```

```python
%%writefile "/content/drive/MyDrive/Smart Gate/Deployment/Dockerfile"
# 1. Base image: official PyTorch image with CUDA and cuDNN
# This is required to run TensorRT and onnxruntime-gpu
FROM pytorch/pytorch:2.1.0-cuda11.8-cudnn8-runtime

# 2. Set the working directory inside the container
WORKDIR /app

# 3. Install system dependencies
# These packages are required because opencv-python needs graphics libraries on Linux
RUN apt-get update && apt-get install -y \
    libgl1-mesa-glx \
    libglib2.0-0 \
    && rm -rf /var/lib/apt/lists/*

# 4. Copy requirements first (to benefit from Docker cache and speed up later builds)
COPY Deployment/requirements.txt .

# 5. Install Python packages
# Use --no-cache-dir to keep the final image smaller
RUN pip install --no-cache-dir --upgrade pip && \
    pip install --no-cache-dir -r requirements.txt

# 6. Copy the project folders into the container
COPY Src/ ./Src/
COPY Weights/ ./Weights/
COPY Deployment/ ./Deployment/

# 7. Set environment variables
# This lets Python import from Src (from Src.models import ...)
ENV PYTHONPATH=/app

# 8. Expose the FastAPI port
EXPOSE 8000

# 9. Default command when the container starts
CMD ["uvicorn", "Deployment.api:app", "--host", "0.0.0.0", "--port", "8000"]
```
