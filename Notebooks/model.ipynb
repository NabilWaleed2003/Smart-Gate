## From PyTorch to ONNX

```python
import torch
from Src.models.Build_model import TruckDetectionModel
from Src.models import optimize

print("1. Building the model skeleton...")

# Here we provide the path to the original YOLO weights
# that were used to initialize the model.
# If you are not sure, use the default model name you started with,
# such as "yolo11m.pt".
base_yolo_weights = "yolo11m.pt"

model = TruckDetectionModel(model_weights_path=base_yolo_weights)

print("2. Loading your best trained weights...")

# Path to the best weights obtained during training
best_weights_path = "/content/drive/MyDrive/Smart Gate/Models/Yolo11m_truck_best.pt"

# Load the trained weights from the saved checkpoint
checkpoint = torch.load(best_weights_path, map_location="cpu")

# Load the trained weights into the model
model.load_state_dict(checkpoint["model_state_dict"])

print("3. Exporting to ONNX...")

# The model is now a fully initialized model object,
# so we can pass it directly to the ONNX export function.
onnx_path = optimize.export_to_onnx(
    trained_model=model,  # Pass the model object, not the weights path
    save_dir="/content/drive/MyDrive/Smart Gate/Weights/",
    model_name="model_optimized.onnx"
)
```

```python
from pathlib import Path
import os
import onnx
from onnxsim import simplify

WEIGHTS_DIR = "/content/drive/MyDrive/Smart Gate/Weights"
onnx_fp32_path = os.path.join(WEIGHTS_DIR, "model_optimized.onnx")
onnx_simplified_path = os.path.join(WEIGHTS_DIR, "model_optimized_sim.onnx")

onnx_model = onnx.load(onnx_fp32_path)


model_simp, check = simplify(onnx_model)


if check:
    print("Simplification successful! Saving simplified model...")
    onnx.save(model_simp, onnx_simplified_path)
else:
    print("Warning: ONNX simplification failed validation, but saving anyway.")
    onnx.save(model_simp, onnx_simplified_path)


print("2. Converting Simplified ONNX to TensorRT...")
optimize.export_onnx_to_tensorrt(
    onnx_model_path=onnx_simplified_path,
    save_dir=WEIGHTS_DIR,
    trt_model_name="model_tensorrt.engine"
)
```

```python
import os

project_path = '/content/drive/MyDrive/Smart Gate'

print(f"Project structure for: {project_path}\n")

# Define folders to exclude (case-insensitive for robustness)
folders_to_exclude = {'raw', 'processed'}

# Define common image file extensions (to identify potential image folders)
image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp'}

for root, dirs, files in os.walk(project_path):
    # Filter out directories for image data
    # Make a copy of dirs to modify it safely during iteration
    dirs[:] = [d for d in dirs if d.lower() not in folders_to_exclude]

    # Also, skip any directory if any file inside it has a common image extension
    # This is a more robust check for 'image folders'
    should_skip_dir = False
    for f in files:
        if os.path.splitext(f)[1].lower() in image_extensions:
            should_skip_dir = True
            break
    if should_skip_dir:
        continue

    level = root.replace(project_path, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f'{indent}{os.path.basename(root)}/')
    subindent = ' ' * 4 * (level + 1)
    for f in files:
        print(f'{subindent}{f}')
```
